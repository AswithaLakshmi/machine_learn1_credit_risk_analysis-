```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

```

```python
train_path = f"{data_path}/application_train.csv"

df = pd.read_csv(train_path)

print("Shape:", df.shape)
```

```python
df["TARGET"].value_counts(normalize=True) * 100
```

```python
missing = df.isnull().sum()

missing[missing > 0].sort_values(ascending=False).head(20)
```

```python
missing = df.isnull().sum()

missing_df = pd.DataFrame({
    "Missing_Count": missing,
    "Missing_Percentage": (missing / len(df)) * 100
})

missing_df = missing_df[missing_df["Missing_Count"] > 0] \
    .sort_values("Missing_Percentage", ascending=False)

missing_df.head(20)
```

```python
total_missing_columns = df.isnull().any().sum()

print("Total columns:", df.shape[1])
print("Columns with missing values:", total_missing_columns)
print("Columns without missing values:", df.shape[1] - total_missing_columns)
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

print("Columns with >50% missing:",
      (missing_percentage > 50).sum())

print("Columns with >30% missing:",
      (missing_percentage > 30).sum())

print("Columns with >10% missing:",
      (missing_percentage > 10).sum())
```

```python
numeric_cols = df.select_dtypes(include=["int64", "float64"]).columns
categorical_cols = df.select_dtypes(include=["object"]).columns

print("Numerical columns:", len(numeric_cols))
print("Categorical columns:", len(categorical_cols))
```

```python
for col in categorical_cols:
    print(f"\n{col}")
    print("Unique values:", df[col].nunique())
    print(df[col].unique())
```
